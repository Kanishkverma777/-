# Heart Disease Dataset - Exploratory Data Analysis (EDA)

This notebook explores the UCI Heart Disease dataset (ID: 45) to analyze feature distributions, missingness patterns, target balance, and feature correlations with heart disease presence.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from heart_disease.config import load_config
from heart_disease.data import load_raw_data, clean_data

sns.set_theme(style="whitegrid")
config = load_config()
raw_df = load_raw_data(config)
print(f"Raw dataset shape: {raw_df.shape}")
raw_df.head()
```

## 1. Missing Value Analysis

```python
missing_counts = raw_df.isnull().sum()
missing_pct = (missing_counts / len(raw_df)) * 100
missing_df = pd.DataFrame({'Missing Count': missing_counts, 'Percentage (%)': missing_pct})
missing_df[missing_df['Missing Count'] > 0]
```

## 2. Target Distribution (Original vs Binarized)

```python
clean_df = clean_data(raw_df, config)
print("Target Class Distribution (Binarized):")
print(clean_df['num'].value_counts(normalize=True) * 100)

fig, ax = plt.subplots(figsize=(6, 4))
sns.countplot(x='num', data=clean_df, palette='Set2', ax=ax)
ax.set_xticklabels(['No Disease (0)', 'Disease (1)'])
ax.set_title('Binarized Target Distribution')
plt.show()
```

## 3. Numeric Feature Distributions by Heart Disease Status

```python
num_cols = config.features.numeric
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.flatten()
for i, col in enumerate(num_cols):
    sns.boxplot(x='num', y=col, data=clean_df, ax=axes[i], palette='Set2')
    axes[i].set_title(f'{col} vs Target')
    axes[i].set_xticklabels(['No Disease', 'Disease'])
plt.tight_layout()
plt.show()
```

## 4. Categorical Feature Association with Disease

```python
cat_cols = ['sex', 'cp', 'fbs', 'exang', 'slope']
fig, axes = plt.subplots(2, 3, figsize=(16, 9))
axes = axes.flatten()
for i, col in enumerate(cat_cols):
    prop_df = clean_df.groupby(col)['num'].mean().reset_index()
    sns.barplot(x=col, y='num', data=prop_df, ax=axes[i], palette='Blues_d')
    axes[i].set_ylabel('Proportion with Disease')
    axes[i].set_title(f'Disease Prevalence by {col}')
axes[-1].set_visible(False)
plt.tight_layout()
plt.show()
```

## 5. Correlation Heatmap of Numeric Features

```python
corr = clean_df[num_cols + ['num']].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1)
plt.title('Correlation Matrix of Numeric Features & Target')
plt.show()
```
